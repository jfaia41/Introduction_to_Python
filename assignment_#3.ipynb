#**Pergunta 2**

Qual é o setor com maior valor mediano de empréstimo em três países específicos?

**Requisitos:**
- Sleção de linhas com `.isin()`;
- Agrupamento por setor;
- Agregação (mediana);
- Ordenação.

```python
# Escolher três países
selected_countries = ["Kenya", "Philippines", "Peru"]

# Filtrar apenas esses países e valores válidos
df_p2 = df.dropna(subset=["loan_amount_usd", "country", "sector"])
df_p2 = df_p2[df_p2["country"].isin(selected_countries)]

# Calcular mediana por país e setor
sector_median = (
    df_p2
    .groupby(["country", "sector"], as_index=False)["loan_amount_usd"]
    .median()
    .rename(columns={"loan_amount_usd": "median_loan_usd"})
)

# Ordenar por país e mediana (descrescente)
sector_median_sorted = sector_median.sort_values(
    ["country", "median_loan_usd"], ascending=[True, False]
)

sector_median_sorted.head(20)
```

# **Pergunta 3**

Entre os empréstimos com intervalo de pagamento `"monthly"`, quais são as 5 atividades que recebem empréstimos de maior valor?

**Requisitos:**
- Filtragem com *strings* (`repayment_interval == "monthly"`);
- Ordenar por ordem descrescente do valor do empréstimo;
- Seleção das top 5 atividades.

```python
# Filtrar empréstimos com pagamento mensal
df_p3 = df.dropna(subset=["loan_amount_usd", "repayment_interval", "activity"])
df_p3 = df_p3[df_p3["repayment_interval"] == "monthly"]

# Calcular valor médio por atividade
activity_mean = (
    df_p3
    .groupby("activity", as_index=False)["loan_amount_usd"]
    .mean()
    .rename(columns={"loan_amount_usd": "mean_loan_usd"})
)

# Ordenar descrescente e mostrar top 5
top5_activities = activity_mean.sort_values("mean_loan_usd", ascending=False).head(5)
top5_activities
```

#**Pergunta 4**

Existe relação entre o número de financiadores por empréstimo e o valor do empréstimo? Criando a variável `“loan_per_lender”`, qual setor apresenta o maior valor médio por financiador?

**Requisitos:**
- Criar uma nova variável com `apply`+ `lambda`:
  - `loan_per_lender = loan_amount_usd / lender_count`;
- Agrupar por setor;
- Agregação (média);
- Ordenar.


```python
# Mostrar as colunas existentes
df.columns.tolist()

# Garantir que as colunas necessárias existem
cols_needed = ["loan_amount_usd", "lender_count", "sector"]
for col in cols_needed:
    if col not in df.columns:
        print(f"Atenção: coluna '{col}' não existe no DataFrame.")

# Filtrar linhas com dados válidos
df_p4 = df.dropna(subset=["loan_amount_usd", "lender_count", "sector"])

# Evitar divisão por zero
df_p4 = df_p4[df_p4["lender_count"] > 0]

# Criar nova variável loan_per_lender com apply + lambda
df_p4["loan_per_lender"] = df_p4.apply(
    lambda row: row["loan_amount_usd"] / row["lender_count"],
    axis=1
)

# Agrupar por setor e calcular média
sector_lpl = (
    df_p4
    .groupby("sector", as_index=False)["loan_per_lender"]
    .mean()
    .rename(columns={"loan_per_lender": "mean_loan_per_lender"})
)

# Ordenar setores do maior para o menor
sector_lpl_sorted = sector_lpl.sort_values("mean_loan_per_lender", ascending=False)
sector_lpl_sorted
```

#**Pergunta 5**

Em que regiões existe a maior proporção de empréstimos atribuídos exclusivamente a mutuárias mulheres (`“female”`)?

**Requisitos cumpridos:**
- Filtragem usando *strings* (`borrower_genders.str.contains("female"))`;
- Agrupamento por região;
- Cáculo de proporção;
- Ordenar.

```python
# Garantir que as colunas existem
cols_needed = ["borrower_genders", "region"]
for col in cols_needed:
    if col not in df.columns:
        print(f"Atenção: coluna '{col}' não existe no DataFrame.")

# Filtrar linhas com região e borrower_genders não nulos
df_p5 = df.dropna(subset=["borrower_genders", "region"])

# Considerar "apenas mulheres" quando:
# - contém "female" e NÃO contém "male"
only_female_mask = df_p5["borrower_genders"].str.contains("female", case=False) & \
                   ~df_p5["borrower_genders"].str.contains("male", case=False)

df_p5["only_female"] = only_female_mask

# Contar total de empréstimos e apenas-femininos por região
region_stats = (
    df_p5
    .groupby("region")
    .agg(
        total_loans=("only_female", "count"),
        female_only_loans=("only_female", "sum")
    )
    .reset_index()
)

# Calcular proporção
region_stats["female_only_ratio"] = region_stats["female_only_loans"] / region_stats["total_loans"]

# Ordenar pela maior proporção
region_stats_sorted = region_stats.sort_values("female_only_ratio", ascending=False)

# Mostrar top 10 regiões
region_stats_sorted.head(10)
```

#**Pergunta 6**

O prazo dos empréstimos (`"loan_term"`) é diferente entre países de alto risco e baixo risco?

**Requisitos:**
- Criação de categorias com condições numéricas;
- Seleção de linhas (alto risco vs. baixo risco);
- Agrupamento e cálculo de média do prazo dos empréstimos;
- Comparação e ordenação.

```python
# Garantir que as colunas existem
cols_needed = ["country", "loan_amount_usd", "term_in_months"]
for col in cols_needed:
    if col not in df.columns:
        print(f"Atenção: coluna '{col}' não existe no DataFrame.")

# Filtrar linhas válidas
df_p6 = df.dropna(subset=["country", "loan_amount_usd", "term_in_months"])

# Calcular valor médio por país
country_mean_p6 = (
    df_p6
    .groupby("country", as_index=False)["loan_amount_usd"]
    .mean()
    .rename(columns={"loan_amount_usd": "mean_loan_usd"})
)

# Calcular quartis
q1 = country_mean_p6["mean_loan_usd"].quantile(0.25)
q3 = country_mean_p6["mean_loan_usd"].quantile(0.75)

# Classificar países por risco
def classify_risk(mean_loan):
    if mean_loan >= q3:
        return "alto_risco"
    elif mean_loan <= q1:
        return "baixo_risco"
    else:
        return "medio"

country_mean_p6["risk_group"] = country_mean_p6["mean_loan_usd"].apply(classify_risk)

# Juntar classificação de risco ao DataFrame original
df_p6 = df_p6.merge(country_mean_p6[["country", "risk_group"]], on="country", how="left")

# Calcular prazo médio por grupo de risco
risk_term = (
    df_p6
    .groupby("risk_group", as_index=False)["term_in_months"]
    .mean()
    .rename(columns={"term_in_months": "mean_term_months"})
)

risk_term
```
